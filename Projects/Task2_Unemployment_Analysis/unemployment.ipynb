```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# ---------- Load Datasets ----------

df1 = pd.read_csv("unemployment in India.csv")
df2 = pd.read_csv("unemployment_Rate_upto_11_2020.csv")

print("Dataset 1 Shape:", df1.shape)
print("Dataset 2 Shape:", df2.shape)

print("\nFirst 5 Rows of Dataset 1:\n")
print(df1.head())

print("\nFirst 5 Rows of Dataset 2:\n")
print(df2.head())
```

```python
# ---------- Basic Cleaning ----------

# Remove extra spaces from column names
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

# Convert Date column to datetime format
df1['Date'] = pd.to_datetime(df1['Date'])
df2['Date'] = pd.to_datetime(df2['Date'])

# Check missing values
print("\nMissing Values in Dataset 1:\n")
print(df1.isnull().sum())

print("\nMissing Values in Dataset 2:\n")
print(df2.isnull().sum())
```

```python
# ------ Remove missing values -------
df1 = df1.dropna()
df2 = df2.dropna()

print("\nMissing Values in Dataset 1:\n")
print(df1.isnull().sum())

print("\nMissing Values in Dataset 2:\n")
print(df2.isnull().sum())
```

```python
# ------- Data Visualization -------

# ---------- 1. Unemployment Rate Distribution ----------

sns.histplot(
    df1['Estimated Unemployment Rate (%)'],
    bins=20,
    kde=True
)

plt.title("Unemployment Rate Distribution")
plt.xlabel("Unemployment Rate (%)")
plt.show()


# ---------- 2. Top 10 States with Highest Unemployment ----------
top_states = df1.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=top_states.index, y=top_states.values)
plt.title("Top 10 States with Highest Unemployment")
plt.xlabel("State")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.show()

# ---------- 3. Urban vs Rural Unemployment ----------
plt.figure(figsize=(6,4))

sns.boxplot(
    data=df1,
    x='Area',
    y='Estimated Unemployment Rate (%)'
)
plt.title("Urban vs Rural Unemployment")
plt.show()


# ---------- 4. Covid Impact on Unemployment ----------
plt.figure(figsize=(12,5))
sns.lineplot(
    data=df2,
    x='Date',
    y='Estimated Unemployment Rate (%)'
)

plt.title("Unemployment Rate During Covid-19")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.show()

# ---------- 5. Region Wise Unemployment ----------
plt.figure(figsize=(10,5))

sns.barplot(
    data=df2,
    x='Region',
    y='Estimated Unemployment Rate (%)'
)

plt.xticks(rotation=90)
plt.title("Region Wise Unemployment Rate")
plt.show()
```

```python
# ------- insights -------

# Average unemployment rate
avg_rate = df1['Estimated Unemployment Rate (%)'].mean()

print("\nAverage Unemployment Rate:",
      round(avg_rate, 2), "%")


# State with highest unemployment
highest_state = top_states.idxmax()

print("State with Highest Unemployment:",
      highest_state)
```

```python
print("\nConclusion:")
print("The unemployment rate increased significantly during Covid-19.")
print("Some states were affected more than others.")
print("Urban and Rural areas also showed different unemployment trends.")
```
